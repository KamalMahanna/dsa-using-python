```python
import numpy as np
```

```python
some_var = np.random.choice(list("abcdefghijklmnopqrstuvwxyz"), 1000000)
```

```python
from collections import Counter
```

```python
%%timeit -n 100
conted_dict = Counter(some_var)
```

```python
%%timeit -n 100

a_dict = {}
for i in some_var:
    if i in a_dict:
        a_dict[i] += 1
    else:
        a_dict[i] = 1
```

```python
%%timeit -n 100

b_dict = {}
for i in some_var:
    b_dict[i] = b_dict.get(i, 0) + 1
```
