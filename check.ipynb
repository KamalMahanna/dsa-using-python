```python
def sortedSquares(nums) :
    pos_idx=len(nums)

    for i in range(len(nums)):
        if nums[i]>=0:
            pos_idx = i
            break

    nums = [i**2 for i in nums]
    if pos_idx==0:
        return nums
    if pos_idx==len(nums):
        return nums[::-1]

    neg_idx = pos_idx -1
    new_list = []
    while pos_idx<len(nums) and neg_idx >-1:
        print(nums[neg_idx],nums[pos_idx],new_list)
        if nums[pos_idx] < nums[neg_idx]:
            new_list.append(nums[pos_idx])
            pos_idx+=1
        elif nums[pos_idx] > nums[neg_idx]:
            new_list.append(nums[neg_idx])
            neg_idx-=1
        else:
            new_list.append(nums[neg_idx])
            new_list.append(nums[neg_idx])
            neg_idx-=1
            pos_idx+=1
    
    while neg_idx>-1:
        new_list.append(nums[neg_idx])
        neg_idx-=1
    while pos_idx<len(nums):
        new_list.append(nums[pos_idx])
        pos_idx+=1
    return new_list
```

```python
def removeDuplicates(nums) -> int:
    a=0
    for i in range(1,len(nums)):
            if nums[a]!=nums[i]:
                a+=1
                nums[a]=nums[i]
    return nums[:a+1]
```

```python
removeDuplicates([1,1,2])
```

```python
removeDuplicates([0,0,1,1,1,2,2,3,3,4])
```

```python
def segregate0and1(arr):
    l=0
    r=len(arr)-1
    while l<r:
        print('l=',l,'r=',r)
        if arr[r]==1:
            r-=1
        elif arr[l]==0:
            l+=1
        elif (arr[r] == 0) and (arr[l]==1):
            arr[r],arr[l] = 1,0
            r-=1
            l+=1
    return arr
```

```python
segregate0and1([0,1,0,1,0,1])
segregate0and1([0,0,0,0,0])
```
